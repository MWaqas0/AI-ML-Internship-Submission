Import Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Load Dataset

```python
iris = sns.load_dataset('iris')

print("Shape:", iris.shape)
print("\nColumns:")
print(iris.columns)

print("\nFirst 5 Rows:")
print(iris.head())
```

Dataset Information

```python
print("\nDataset Info:")
print(iris.info())

print("\nStatistics:")
print(iris.describe())
```

Scatter Plot

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    x='sepal_length',
    y='petal_length',
    hue='species',
    data=iris
)

plt.title("Scatter Plot")
plt.show()
```

Histograms

```python
iris.hist(figsize=(10,8))

plt.tight_layout()
plt.show()
```

Boxplot

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=iris)

plt.title("Boxplot for Outlier Detection")

plt.show()
```

Final Insights

```python
print("""
Insights:

1. Iris dataset has 3 species.
2. Petal features separate classes better.
3. Few outliers exist.
4. Dataset balanced appears.
""")
```
