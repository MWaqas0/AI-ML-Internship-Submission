Import Libraries

```python
from transformers import pipeline
```

Load Model

```python
chatbot = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

Safety Function

```python
def safety_filter(question):

    restricted_words = [
        "suicide",
        "overdose",
        "kill",
        "harm"
    ]

    for word in restricted_words:

        if word in question.lower():

            return False

    return True
```

Prompt Engineering

```python
def ask_health_bot(question):

    if not safety_filter(question):

        return "I cannot provide unsafe medical advice. Please consult a professional."

    prompt = f"""
    Act like a helpful medical assistant.

    Give friendly and clear information.

    Question: {question}

    Answer:
    """

    response = chatbot(
        prompt,
        max_length=120,
        do_sample=True
    )

    return response[0]["generated_text"]
```
